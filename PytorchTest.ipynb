```python
import torch.nn as nn
import torch.nn.functional as F
```

```python
#create a model class that inherits nn.module
class Model(nn.Module):
  #input layer (4 features of the flower)-->
  #hidden layer1(number of neurons)--> H2(n)-->output
  def __init__(self,in_features=4,h1=8,h2=9,out_features=3):
    super().__init__()
    self.fc1 = nn.Linear(in_features,h1)
    self.fc2 = nn.Linear(h1,h2)
    self.out = nn.Linear(h2,out_features)

  def forward(self,x):
    x = F.relu(self.fc1(x))
    x = F.relu(self.fc2(x))
    x = self.out(x)

    return x
```

```python
#pick a manual seed for randomisaion
torch.manual_seed(41)
#create instance of model
model = Model()
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
import pandas as pd
```

```python
url = 'https://gist.githubusercontent.com/netj/8836201/raw/6f9306ad21398ea43cba4f7d537619d0e07d5ae3/iris.csv'
my_df = pd.read_csv(url)
```

```python
my_df
```

```python
#update last column from string to integer
my_df ['variety'] = my_df['variety'].replace('Setosa',0.0)
my_df ['variety'] = my_df['variety'].replace('Versicolor',1.0)
my_df ['variety'] = my_df['variety'].replace('Virginica',2.0)
my_df
```

```python
#train test split x,7
x = my_df.drop('variety',axis=1)
y = my_df['variety']
```

```python
#convert to  numpy arrays
x
```

```python
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=41)

# Convert to PyTorch tensors
X_train = torch.from_numpy(X_train).float()
X_test = torch.from_numpy(X_test).float()
y_train = torch.from_numpy(y_train).long()
y_test = torch.from_numpy(y_test).long()
```

```python
#set criteria to measure error
criterion = nn.CrossEntropyLoss()
#choose adam optimiser , lr = learning rate(in error doesnt go down after a bunch of errors, we lower the rate)
optimizer = torch.optim.Adam(model.parameters(),lr=0.01)
```

```python
#train our model
#epochs?(one run thru all the training data in the network)
epochs = 1000
losses = []
for i in range(epochs):
  #go get a prediction
  y_pred = model.forward(X_train) #get predicted results

  #measure the loss/error
  loss = criterion(y_pred,y_train) #compare predicted

  #keep track of our losses
  losses.append(loss.detach().numpy())

  #print every 10 epoch
  if i % 10 == 0:
    print(f'Epoch:{i} Loss:{loss}')

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

#plot our losses
plt.plot(range(epochs),losses)
```

```python
import torch
import math

def sinosiodal_positional_encoding(seq_len, d_model):
  PE = torch.zeros(seq_len, d_model)
  for pos in range(seq_len):
    for i in range(0, d_model, 2):
      PE[pos, i] = math.sin(pos / (10000 ** (i/ d_model)))
      if i + 1 < d_model:
        PE[pos, i + 1] = math.cos(pos / (10000 ** (i/ d_model)))
  return PE

```

```python
import torch
import math

def build_rope_cache(seq_len, dim, base=10000):
  """
  def creates lookup table
  dim becoz of the dimension of the head(must be even)
  base is frequency
  """
  assert dim % 2 == 0, "dim must be an even integer."
  #compute frequency
  inv_freq = 1.0 / (base ** (torch.arange(0, dim, 2).float() / dim))
  #positions
  positions = torch.arange(seq_len).float()
  #outer product (seq_len , dim/2)
  freqs = torch.outer(positions, inv_freq)
  #precompute sin $ cos
  sin= freqs.sin() #(seq_len , dim/2)
  cos = freqs.cos() #(seq_len , dim/2)

  return sin,cos
  #PE = torch.zeros(seq_len, dim)
```

```python
def apply_rope_interleaved(x, sin, cos):
  """
  x: (batch, heads, seq_len, head_dim)
  sin, cos: (seq_len, head_dim/2)
  """
  b, h, seq_len, d = x.shape
  assert d % 2 == 0, "d must be an even integer."

  sin = sin[:seq_len].to(dtype=x.dtype, device=x.device)
  cos = cos[:seq_len].to(dtype=x.dtype, device=x.device)

  #expand for broadcasting
  sin = sin.unsqueeze(0).unsqueeze(0) #(1,1, seq_len, d/2)
  cos = cos.unsqueeze(0).unsqueeze(0) #(1,1, seq_len, d/2)

  #reshape to pairs (can use x.view or x.reshape)
  x.view(b,h, seq_len, d // 2, 2)

  x_even = x[..., 0]
  x_odd = x[..., 1]

  #apply rotation
  rotated_even = x_even * cos - x_odd * sin #shape: (b,h, seq_len, d / 2)
  rotated_odd = x_even * sin + x_odd * cos #shape: (b,h, seq_len, d / 2 )

  #restack
  x = torch.stack((rotated_even, rotated_odd), dim=-1) #shape: (b,h, seq_len, d / 2 ,2 )
  return x.view(b, h, seq_len, d) #shape
```
